```python
raw_times = ["112ms", "998ms", "101ms", "115ms", "1904ms"]

response_times = []
for t in raw_times:
    cleaned = int(t.rstrip("ms"))
    response_times.append(cleaned)

response_times
```

```python
response_times = [int(t.rstrip("ms")) for t in raw_times]
response_times
```

```python
high_response_times = [t for t in response_times if t > 500]
high_response_times
```

```python
high_response_times = []
for t in response_times:
  if t > 500:
    high_response_times.append(t)

high_response_times
```

```python
# Dictionary comprehension
```

```python
cities = ["London", "New York", "Berlin"]

city_name_length = {city : len(city) for city in cities}
city_name_length
```

```python
records = [
    {"id": 1001, "city": "London", "temp": 21.7},
    {"id": 1002, "city": "New York", "temp": 19.3},
    {"id": 1003, "city": "Berlin", "temp": 20.1},
]

temp_for_city = {}
for row in records:
    temp_for_city[row["city"]] = row["temp"]
temp_for_city
```

```python
temp_for_city = { row["city"] : row["temp"] for row in records}
```

```python
records = [
    {"user": "bob",   "page": "/"},
    {"user": "alice", "page": "/contact"},
    {"user": "carol", "page": "/search"},
    {"user": "bob",   "page": "/pricing"},
    {"user": "bob",   "page": "/contact"},
]

visited_pages_list = [record["page"] for record in records]
visited_pages_list
```

```python
visited_pages_set = {record["page"] for record in records}
visited_pages_set
```
